## 🔹 **What are Ensemble Methods?**
Ensemble methods are **machine learning techniques** that combine multiple models to improve predictive performance. The goal is to create a **stronger model** that performs better than a single weak classifier. Ensemble methods help to:
- Reduce **overfitting**.
- Increase **robustness**.
- Improve **accuracy and stability**.

Two popular ensemble methods are **Random Forest** and **AdaBoost**.

---

## 🌳 **Random Forest**
Random Forest is an ensemble learning method that builds multiple **decision trees** and combines their predictions to improve accuracy.

### **How Random Forest Works?**
1. **Bootstrapping:** Creates multiple random subsets of the dataset (bagging).
2. **Decision Trees:** Builds multiple decision trees on these subsets.
3. **Aggregation:** Uses **majority voting** (classification) or **averaging** (regression) to make predictions.

### **Advantages**
✅ Handles large datasets well.  
✅ Works with both categorical and numerical data.  
✅ Less prone to overfitting compared to individual decision trees.  
✅ Reduces variance by combining multiple models.  

---

## ⚡ **AdaBoost (Adaptive Boosting)**
AdaBoost is a boosting algorithm that **sequentially trains weak classifiers**, giving more focus to incorrectly classified samples in each iteration.

### **How AdaBoost Works?**
1. **Assign Equal Weights**: Each sample starts with equal importance.
2. **Train Weak Learner**: Usually a simple decision tree (stump).
3. **Update Weights**: Misclassified points get **higher** weights, making them more important for the next classifier.
4. **Combine Weak Learners**: Final prediction is a **weighted sum** of all weak models.

### **Advantages**
✅ Can boost weak models into strong models.  
✅ Less prone to overfitting than a single decision tree.  

---

## 🔍 **Comparison: Random Forest vs. AdaBoost**
| Feature         | Random Forest | AdaBoost |
|----------------|--------------|----------|
| **Base Learner** | Multiple decision trees | Weak learners (e.g., decision stumps) |
| **Focus** | Reduces variance (bagging) | Reduces bias (boosting) |
| **Handling Noise** | Works well with noisy data | Sensitive to noise |
| **Training Speed** | Faster (trees trained in parallel) | Slower (sequential training) |
| **Overfitting** | Less prone to overfitting | Can overfit on noisy datasets |

---

## **🚀 Summary**
- **Random Forest**: Best for **reducing variance** and handling large datasets.
- **AdaBoost**: Best for **reducing bias**, but can be sensitive to noise.
